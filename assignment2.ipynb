```sql
--Q1)find the total oilproduced_bbl and total gasproduced_mcf for each field ,but only include 'active' wells.

SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_Oil,
    SUM(GasProduced_mcf) AS Total_Gas
FROM well_production
WHERE Status = 'Active'
GROUP BY Field;
```

```sql
--Q2. Find the average OperatingCost for each OperatorName, sorted from highest to lowest.

SELECT
    OperatorName,
    AVG(OperatingCost) AS Average_OperatingCost
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OperatingCost DESC;
```

```sql
--Q3. Find the number of wells per Region, but only for regions that have more than 2 wells.

SELECT
    Region,
    COUNT(*) AS Total_Wells
FROM well_production
GROUP BY Region
HAVING COUNT(*) > 2;
```

```sql
--Q4)Display the top 3 Fields with the highest total OilProduced_bbl.

SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_Oil
FROM well_production
GROUP BY Field
ORDER BY Total_Oil DESC
LIMIT 3;
```

```sql
--Q5. Find the minimum and maximum DepthMeters for each WellType, but only consider wells that are NOT Abandoned.

SELECT
    WellType,
    MIN(DepthMeters) AS Minimum_Depth,
    MAX(DepthMeters) AS Maximum_Depth
FROM well_production
WHERE Status <> 'Abandoned'
GROUP BY WellType;
```

```sql
--Q6. List each unique OperatorName along with the total number of wells and average OilProduced_bbl, sorted by average oil production (highest first), showing only the top 5.

SELECT
    OperatorName,
    COUNT(*) AS Total_Wells,
    AVG(OilProduced_bbl) AS Average_OilProduced
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OilProduced DESC
LIMIT 5;
```
