```sql
-- ============================================================
-- Q1. Rank all watches by price in descending order
--     using RANK().
-- ============================================================

SELECT model_name,
       price,
       RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql
-- Q2. Rank all watches using DENSE_RANK().
-- ============================================================

SELECT model_name,
       price,
       DENSE_RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql
-- Q3. Assign a unique sequential number to every sale
--     ordered by sale_date, oldest first.
-- ============================================================

SELECT sale_id,
       model_name,
       sale_date,
       ROW_NUMBER() OVER (ORDER BY sale_date ASC) AS sale_number
FROM watch_sales;
```

```sql
-- Q4. Within each category, rank watches by units_sold
--     from highest to lowest using RANK().
-- ============================================================

SELECT model_name,
       category,
       units_sold,
       RANK() OVER (
           PARTITION BY category
           ORDER BY units_sold DESC
       ) AS category_rank
FROM watch_sales;
```

```sql
-- Q5. Find the top 3 best-selling watches in each category
--     using DENSE_RANK() inside a CTE.
-- ============================================================

WITH RankedWatches AS
(
    SELECT model_name,
           category,
           units_sold,
           DENSE_RANK() OVER (
               PARTITION BY category
               ORDER BY units_sold DESC
           ) AS sales_rank
    FROM watch_sales
)

SELECT model_name,
       category,
       units_sold,
       sales_rank
FROM RankedWatches
WHERE sales_rank <= 3;
```

```sql
-- Q6. Use ROW_NUMBER() to identify and list only the
--     most recent sale for each brand.
-- ============================================================

WITH RecentSales AS
(
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY brand_name
               ORDER BY sale_date DESC
           ) AS row_num
    FROM watch_sales
)

SELECT *
FROM RecentSales
WHERE row_num = 1;
```

```sql
-- Q8. Use CONCAT() to build a full_name column from
--     customer_first_name and customer_last_name.
-- ============================================================

SELECT CONCAT(
           customer_first_name,
           ' ',
           customer_last_name
       ) AS full_name
FROM watch_sales;
```

```sql
-- Q9. Use TRIM() to remove leading and trailing spaces
--     from brand_name.
-- ============================================================

SELECT brand_name,
       TRIM(brand_name) AS cleaned_brand
FROM watch_sales;
```

```sql
-- Q10. Extract the first 4 characters of model_name
--      using SUBSTRING().
-- ============================================================

SELECT model_name,
       SUBSTRING(model_name, 1, 4) AS first_four_characters
FROM watch_sales;
```

```sql
-- Q11. Replace all hyphens in model_name with spaces.
-- ============================================================

SELECT model_name,
       REPLACE(model_name, '-', ' ') AS cleaned_model
FROM watch_sales;
```

```sql
-- Q12. Standardize country to uppercase using UPPER().
-- ============================================================

SELECT country,
       UPPER(country) AS uppercase_country
FROM watch_sales;
```

```sql
-- Q12  Standardize country to lowercase
--      using LOWER().
-- ============================================================

SELECT country,
       LOWER(country) AS lowercase_country
FROM watch_sales;
```

```sql
-- Q13. Combine TRIM() and UPPER() to clean and standardize
--      brand_name.
-- ============================================================

SELECT brand_name,
       UPPER(TRIM(brand_name)) AS cleaned_brand
FROM watch_sales;

```

```sql
-- Q14. Clean first name and last name using TRIM()
--      and proper case.
--      Combine them into full_name.
--      Rank customers by total units_sold using DENSE_RANK().
-- ============================================================
WITH CustomerSales AS
(
    SELECT
        CONCAT(
            INITCAP(LOWER(TRIM(customer_first_name))),
            ' ',
            INITCAP(LOWER(TRIM(customer_last_name)))
        ) AS full_name,
        SUM(units_sold) AS total_units_sold
    FROM watch_sales
    GROUP BY
        customer_first_name,
        customer_last_name
),

RankedCustomers AS
(
    SELECT
        full_name,
        total_units_sold,
        DENSE_RANK() OVER (
            ORDER BY total_units_sold DESC
        ) AS sales_rank
    FROM CustomerSales
)
SELECT *
FROM RankedCustomers
ORDER BY sales_rank;

```

```sql
-- Q15. Mask customer_last_name so only the first letter
--      is visible and the remaining letters are '*'.
--
--      Example:
--      KUMAR -> K****
--
--      Then use ROW_NUMBER() to number the list
--      alphabetically by customer_last_name.
-- ============================================================

SELECT
    ROW_NUMBER() OVER (
        ORDER BY customer_last_name
    ) AS row_num,

    customer_first_name,

    CONCAT(
        SUBSTRING(customer_last_name, 1, 1),
        REPEAT('*', LENGTH(customer_last_name) - 1)
    ) AS masked_last_name

FROM watch_sales
ORDER BY customer_last_name;
```
