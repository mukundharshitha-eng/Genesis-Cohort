```sql
-- 1. CURRENT_DATE
-- Display customer_name, order_date, and today's date.

SELECT
    customer_name,
    order_date,
    CURRENT_DATE() AS today
FROM audio_sales;
```

```sql
-- 2. DATEDIFF
-- Calculate the number of days between order_date
-- and delivery_date.

SELECT
    order_id,
    customer_name,
    order_date,
    delivery_date,
    DATEDIFF(delivery_date, order_date) AS delivery_days
FROM audio_sales;
```

```sql
-- 3. DATE_ADD
-- Find the date that is 7 days after each order_date.

SELECT
    order_id,
    customer_name,
    order_date,
    DATE_ADD(order_date, 7) AS date_after_7_days
FROM audio_sales;
```

```sql
-- 4. EXTRACT
-- Extract the year, month, and day from order_date.

SELECT
    order_id,
    order_date,
    EXTRACT(YEAR FROM order_date) AS order_year,
    EXTRACT(MONTH FROM order_date) AS order_month,
    EXTRACT(DAY FROM order_date) AS order_day
FROM audio_sales;
```

```sql
-- 5. CASE WHEN
-- Create product_type:
-- AirPods   → Apple Product
-- Earphones → Other Brand

SELECT
    product_name,
    category,
    CASE
        WHEN category = 'AirPods' THEN 'Apple Product'
        WHEN category = 'Earphones' THEN 'Other Brand'
        ELSE 'Unknown'
    END AS product_type
FROM audio_sales;
```

```sql
-- 7. COALESCE
-- Replace NULL values in discount with 0.

SELECT
    order_id,
    discount,
    COALESCE(discount, 0) AS final_discount
FROM audio_sales;
```

```sql
-- 8. COALESCE
-- Replace NULL values in rating with 0.

SELECT
    order_id,
    rating,
    COALESCE(rating, 0) AS rating_value
FROM audio_sales;
```

```sql
-- 9. NULLIF
-- Convert a rating of 0 into NULL.

SELECT
    order_id,
    rating,
    NULLIF(rating, 0) AS converted_rating
FROM audio_sales;
```

```sql
-- 10. NULLIF
-- Calculate discount percentage:
-- discount / price × 100
-- NULLIF prevents division by zero.

SELECT
    order_id,
    price,
    discount,
    (discount / NULLIF(price, 0)) * 100 AS discount_percentage
FROM audio_sales;
```

```sql
-- 11. COMBINED
-- Find delivery days and create a category:
-- 3 days or less → Fast
-- 4–5 days       → Normal
-- More than 5    → Slow

SELECT
    order_id,
    order_date,
    delivery_date,

    DATEDIFF(delivery_date, order_date) AS delivery_days,

    CASE
        WHEN DATEDIFF(delivery_date, order_date) <= 3
            THEN 'Fast'
        WHEN DATEDIFF(delivery_date, order_date) BETWEEN 4 AND 5
            THEN 'Normal'
        ELSE 'Slow'
    END AS delivery_category

FROM audio_sales;
```

```sql
-- 12. COMBINED
-- Display final discount using COALESCE,
-- calculate sales using price × quantity,
-- and classify sales using CASE WHEN.

SELECT
    order_id,
    product_name,
    price,
    quantity,

    COALESCE(discount, 0) AS final_discount,

    price * quantity AS sales,

    CASE
        WHEN price * quantity >= 500 THEN 'High Sales'
        WHEN price * quantity >= 200 THEN 'Medium Sales'
        ELSE 'Low Sales'
    END AS sales_category

FROM audio_sales;
```

```sql
-- 13. BONUS
-- Use at least 5 functions:
-- CURRENT_DATE
-- DATEDIFF
-- DATE_ADD
-- EXTRACT
-- COALESCE
-- NULLIF

SELECT
    order_id,
    customer_name,

    CURRENT_DATE() AS today,

    DATEDIFF(delivery_date, order_date) AS delivery_days,

    DATE_ADD(order_date, 7) AS seven_days_after_order,

    EXTRACT(YEAR FROM order_date) AS order_year,

    COALESCE(discount, 0) AS final_discount,

    NULLIF(rating, 0) AS rating_value

FROM audio_sales;
```
