```sql
CREATE TABLE Laptop_Sales (
    LaptopID INT,
    Brand VARCHAR(50),
    Model VARCHAR(50),
    Price DECIMAL(10,2),
    QuantitySold INT,
    Store VARCHAR(50),
    SaleDate DATE
);

INSERT INTO Laptop_Sales VALUES
(101,'Dell','Inspiron',650,15,'Dallas','2024-01-10'),
(102,'HP','Pavilion',750,20,'Dallas','2024-01-12'),
(103,'Lenovo','ThinkPad',900,10,'Austin','2024-01-15'),
(104,'Apple','MacBook Air',1200,8,'Austin','2024-01-18'),
(105,'Dell','XPS',1400,5,'Houston','2024-01-20'),
(106,'HP','EliteBook',1100,12,'Houston','2024-01-22'),
(107,'Lenovo','IdeaPad',700,18,'Dallas','2024-01-25'),
(108,'Apple','MacBook Pro',1800,6,'Austin','2024-01-28');
select*from Laptop_Sales;
```

```sql
-- 1. Find all laptops that cost more than the average
--    laptop price.


SELECT *
FROM Laptop_Sales
WHERE Price >
(
    SELECT AVG(Price)
    FROM Laptop_Sales
);
```

```sql
-- 2. Find the most expensive laptop.


SELECT *
FROM Laptop_Sales
WHERE Price =
(
    SELECT MAX(Price)
    FROM Laptop_Sales
);
```

```sql
-- 3. Find the cheapest laptop.


SELECT *
FROM Laptop_Sales
WHERE Price =
(
    SELECT MIN(Price)
    FROM Laptop_Sales
);
```

```sql
-- 4. Find laptops sold in the same store as the
--    MacBook Air.


SELECT *
FROM Laptop_Sales
WHERE Store =
(
    SELECT Store
    FROM Laptop_Sales
    WHERE Model = 'MacBook Air'
);
```

```sql
-- 5. Find laptops whose quantity sold is greater
--    than the average quantity sold.


SELECT *
FROM Laptop_Sales
WHERE QuantitySold >
(
    SELECT AVG(QuantitySold)
    FROM Laptop_Sales
);
```

```sql
-- 6. Find laptops sold on the latest sale date.


SELECT *
FROM Laptop_Sales
WHERE SaleDate =
(
    SELECT MAX(SaleDate)
    FROM Laptop_Sales
);
```

```sql
-- 7. Find laptops sold after the earliest sale date.


SELECT *
FROM Laptop_Sales
WHERE SaleDate >
(
    SELECT MIN(SaleDate)
    FROM Laptop_Sales
);
```

```sql
-- 8. Find the brand(s) with the highest total
--    quantity sold.


SELECT Brand,
       SUM(QuantitySold) AS TotalSold
FROM Laptop_Sales
GROUP BY Brand
HAVING SUM(QuantitySold) =
(
    SELECT MAX(TotalSold)
    FROM
    (
        SELECT SUM(QuantitySold) AS TotalSold
        FROM Laptop_Sales
        GROUP BY Brand
    ) AS BrandSales
);
```

```sql
-- 9. Find laptops that cost more than every Dell laptop.

SELECT *
FROM Laptop_Sales
WHERE Price >
(
    SELECT MAX(Price)
    FROM Laptop_Sales
    WHERE Brand = 'Dell'
);
```

```sql
-- 10. Find laptops whose price matches the price of any Apple laptop.

SELECT *
FROM Laptop_Sales
WHERE Price IN
(
    SELECT Price
    FROM Laptop_Sales
    WHERE Brand = 'Apple'
);
```
