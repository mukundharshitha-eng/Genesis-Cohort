```sql
-- TABLE 2 : AGENTS
-- ============================================

CREATE TABLE Agents (
    agent_id INT PRIMARY KEY,
    agent_name VARCHAR(50),
    office_city VARCHAR(50)
);

INSERT INTO Agents VALUES
(101,'David','Dallas'),
(102,'Sarah','Austin'),
(103,'Robert','Houston');
SELECT*FROM Agents;
```

```sql
-- TABLE 3 : PROPERTIES
-- ============================================

CREATE TABLE Properties (
    property_id INT PRIMARY KEY,
    property_name VARCHAR(100),
    city VARCHAR(50),
    property_type VARCHAR(30),
    price INT,
    agent_id INT
);

INSERT INTO Properties VALUES
(1001,'Green Villa','Dallas','Villa',450000,101),
(1002,'Lake House','Austin','House',620000,102),
(1003,'Sky Apartments','Houston','Apartment',380000,103),
(1004,'Oak Residency','Dallas','House',520000,101),
(1005,'River View','Austin','Villa',700000,102),
(1006,'Palm Residency','Houston','House',490000,103);
select*from Properties;
```

```sql
-- 1. Display the customer name, property name,
--    property price, and sale date.
-- ============================================

SELECT c.customer_name,
       p.property_name,
       p.price,
       s.sale_date
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id;
```

```sql
-- 2. Display every property along with the
--    agent who manages it.
-- ============================================

SELECT p.property_name,
       p.property_type,
       p.city,
       p.price,
       a.agent_name
FROM Properties p
INNER JOIN Agents a
ON p.agent_id = a.agent_id;
```

```sql
-- 3. Show customer name, property name,
--    and agent name for every sale.
-- ============================================

SELECT c.customer_name,
       p.property_name,
       a.agent_name
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
INNER JOIN Agents a
ON p.agent_id = a.agent_id;
```

```sql
-- 4. Find all customers who purchased
--    a property worth more than $500,000.
-- ============================================

SELECT c.customer_name,
       p.property_name,
       p.price
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
WHERE p.price > 500000;
```

```sql
-- 5. Show all properties sold by
--    agent 'Sarah'.
-- ============================================

SELECT p.property_name,
       p.price,
       a.agent_name
FROM Properties p
INNER JOIN Agents a
ON p.agent_id = a.agent_id
WHERE a.agent_name = 'Sarah';
```

```sql
-- 6. Display customers who purchased
--    properties located in the same city
--    where they live.
-- ============================================

SELECT c.customer_name,
       c.city,
       p.property_name,
       p.city AS property_city
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
WHERE c.city = p.city;
```

```sql
-- 7. Find the total sales value
--    handled by each agent.
-- ============================================

SELECT a.agent_name,
       SUM(p.price) AS total_sales_value
FROM Agents a
INNER JOIN Properties p
ON a.agent_id = p.agent_id
INNER JOIN Sales s
ON p.property_id = s.property_id
GROUP BY a.agent_name;
```

```sql
-- 8. Count how many properties
--    each customer purchased.
-- ============================================

SELECT c.customer_name,
       COUNT(s.property_id) AS total_properties
FROM Customers c
LEFT JOIN Sales s
ON c.customer_id = s.customer_id
GROUP BY c.customer_name;
```

```sql
-- 9. Find the most expensive property
--    sold by each agent.
-- ============================================

SELECT a.agent_name,
       MAX(p.price) AS highest_property_price
FROM Agents a
INNER JOIN Properties p
ON a.agent_id = p.agent_id
INNER JOIN Sales s
ON p.property_id = s.property_id
GROUP BY a.agent_name;
```

```sql
-- 10. Generate a complete sales report.
-- ============================================

SELECT s.sale_id,
       c.customer_name,
       c.city AS customer_city,
       p.property_name,
       p.property_type,
       p.city AS property_city,
       p.price,
       a.agent_name,
       a.office_city,
       s.sale_date
FROM Sales s
INNER JOIN Customers c
ON s.customer_id = c.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
INNER JOIN Agents a
ON p.agent_id = a.agent_id
ORDER BY s.sale_date;
```
