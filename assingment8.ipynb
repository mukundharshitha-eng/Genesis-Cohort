```sql
-- 1. Find all policies where the premium is greater than
--    the average premium.
-- ============================================================

WITH AvgPremium AS
(
    SELECT AVG(Premium) AS Avg_Premium
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE Premium >
(
    SELECT Avg_Premium
    FROM AvgPremium
);
```

```sql
-- 2. Find the customer(s) with the highest policy amount.
-- ============================================================

WITH MaxPolicy AS
(
    SELECT MAX(PolicyAmount) AS MaxAmount
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE PolicyAmount =
(
    SELECT MaxAmount
    FROM MaxPolicy
);
```

```sql
-- 3. Find the customer(s) with the highest claim amount.
-- ============================================================

WITH MaxClaim AS
(
    SELECT MAX(ClaimAmount) AS MaxClaimAmount
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE ClaimAmount =
(
    SELECT MaxClaimAmount
    FROM MaxClaim
);
```

```sql
-- 4. Calculate the total premium collected from each port.
-- ============================================================

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortPremium;
```

```sql
-- 5. Find ports where the total premium collected is
--    greater than 15000.
-- ============================================================

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortPremium
WHERE TotalPremium > 15000;
```

```sql
-- 6. Calculate the total policy amount for each cargo type.
-- ============================================================

WITH CargoPolicy AS
(
    SELECT CargoType,
           SUM(PolicyAmount) AS TotalPolicy
    FROM MarineInsurance
    GROUP BY CargoType
)

SELECT *
FROM CargoPolicy;
```

```sql
-- 7. Find the average policy amount for each port.
-- ============================================================

WITH AvgPolicy AS
(
    SELECT Port,
           AVG(PolicyAmount) AS AvgPolicyAmount
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM AvgPolicy;
```

```sql
-- 8. Find policies whose policy amount is greater than
--    the average policy amount.
-- ============================================================

WITH AvgPolicy AS
(
    SELECT AVG(PolicyAmount) AS AvgAmount
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE PolicyAmount >
(
    SELECT AvgAmount
    FROM AvgPolicy
);
```

```sql
-- 9. Find the port with the highest total policy amount.
-- ============================================================

WITH PortPolicy AS
(
    SELECT Port,
           SUM(PolicyAmount) AS TotalPolicy
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortPolicy
WHERE TotalPolicy =
(
    SELECT MAX(TotalPolicy)
    FROM PortPolicy
);
```

```sql
-- 10. Find policies whose claim amount is greater than
--     the average claim amount.
-- ============================================================

WITH AvgClaim AS
(
    SELECT AVG(ClaimAmount) AS AvgClaim
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE ClaimAmount >
(
    SELECT AvgClaim
    FROM AvgClaim
);
```

```sql
-- 11. Find policies where the policy amount is greater than
--     the average policy amount of their own port.
-- ============================================================

WITH PortAverage AS
(
    SELECT Port,
           AVG(PolicyAmount) AS AvgPolicy
    FROM MarineInsurance
    GROUP BY Port
)

SELECT m.*
FROM MarineInsurance m
JOIN PortAverage p
ON m.Port = p.Port
WHERE m.PolicyAmount > p.AvgPolicy;
```

```sql
-- 12. Find cargo types whose total claim amount exceeds
--     100000.
-- ============================================================

WITH CargoClaim AS
(
    SELECT CargoType,
           SUM(ClaimAmount) AS TotalClaim
    FROM MarineInsurance
    GROUP BY CargoType
)

SELECT *
FROM CargoClaim
WHERE TotalClaim > 100000;
```

```sql
-- 13. Find customers whose premium is greater than
--     the average premium of their port.
-- ============================================================

WITH PortPremiumAvg AS
(
    SELECT Port,
           AVG(Premium) AS AvgPremium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT m.CustomerName,
       m.Port,
       m.Premium
FROM MarineInsurance m
JOIN PortPremiumAvg p
ON m.Port = p.Port
WHERE m.Premium > p.AvgPremium;
```

```sql
-- 14. Find policies that contribute more than 25%
--     of their port's total premium.
-- ============================================================

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS TotalPremium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT m.PolicyID,
       m.CustomerName,
       m.Port,
       m.Premium
FROM MarineInsurance m
JOIN PortPremium p
ON m.Port = p.Port
WHERE m.Premium > (0.25 * p.TotalPremium);
```

```sql
-- 15. Calculate the total premium, total claim amount,
--     and total policy amount for each port.
-- ============================================================

WITH PortSummary AS
(
    SELECT Port,
           SUM(Premium) AS TotalPremium,
           SUM(ClaimAmount) AS TotalClaimAmount,
           SUM(PolicyAmount) AS TotalPolicyAmount
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortSummary;
```
