```sql
-- Q3. CASE WHEN
-- Sales_Amount >= 1500 → High
-- Sales_Amount >= 800  → Medium
-- Otherwise            → Low
-- ============================================================

SELECT
    Sale_ID,
    Customer_Name,
    Sales_Amount,
    CASE
        WHEN Sales_Amount >= 1500 THEN 'High'
        WHEN Sales_Amount >= 800 THEN 'Medium'
        ELSE 'Low'
    END AS Sales_Category
FROM sales;
```

```sql
-- Q4. CASE WHEN
-- Completed → Successful
-- Cancelled → Unsuccessful
-- ============================================================

SELECT
    Sale_ID,
    Customer_Name,
    Status,
    CASE
        WHEN Status = 'Completed' THEN 'Successful'
        WHEN Status = 'Cancelled' THEN 'Unsuccessful'
    END AS Order_Result
FROM sales;
```

```sql
-- Q5. LAG()
-- For each customer, display previous month's sales.
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales

FROM sales;
```

```sql
-- Q6. LEAD()
-- For each customer, display next month's sales.
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    LEAD(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Next_Month_Sales

FROM sales;
```

```sql
-- Q7. MONTH-TO-MONTH SALES CHANGE
-- Current Sales - Previous Month Sales
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    Sales_Amount -
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Sales_Change

FROM sales;
```

```sql
-- Q8. MONTH-TO-MONTH PERCENTAGE
--
-- (Current Sales - Previous Sales)
-- / Previous Sales * 100
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    (
        Sales_Amount -
        LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        )
    )
    /
    NULLIF(
        LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        ), 0
    ) * 100 AS Percentage_Change

FROM sales;
```

```sql
-- Q9. SALES TREND
--
-- No previous sale → First Month
-- Current > Previous → Increased
-- Current < Previous → Decreased
-- Current = Previous → No Change
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    CASE
        WHEN LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        ) IS NULL
            THEN 'First Month'
         WHEN Sales_Amount > LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        )
            THEN 'Increased'

        WHEN Sales_Amount < LAG(Sales_Amount) OVER (
            PARTITION BY Customer_Name
            ORDER BY Sale_Date
        )
            THEN 'Decreased'

        ELSE 'No Change'
    END AS Sales_Trend

FROM sales;
```

```sql
-- Q10. LAG() + LEAD()
-- Display previous and next month's sales.
-- ============================================================

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,

    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales,

    LEAD(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Next_Month_Sales

FROM sales;
```
